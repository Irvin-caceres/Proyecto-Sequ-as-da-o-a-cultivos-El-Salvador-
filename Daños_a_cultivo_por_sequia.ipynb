```python
# Datos específicos de El Salvador
print("\n ÁREA DE ESTUDIO: REPÚBLICA DE EL SALVADOR")
print("• Regiones: 14 departamentos")
print("• Cultivos principales: Maíz, frijol, sorgo, café")
print("• Temporadas: Primera (Mayo-Julio), Postrera (Agosto-Octubre)")
```

```python
#Precipitación por departamento
print("\n DATOS DE PRECIPITACIÓN POR DEPARTAMENTO (2024-2025)")

departamentos = [
    "Ahuachapán", "Santa Ana", "Sonsonate", "Chalatenango", "La Libertad",
    "San Salvador", "Cuscatlán", "La Paz", "Cabañas", "San Vicente",
    "Usulután", "San Miguel", "Morazán", "La Unión"
]
```

```python
# Precipitación anual promedio (mm) y desviación
precipitacion_promedio = [1800, 1900, 1700, 1600, 1500, 1700, 1600, 1400, 1300, 1500, 1600, 1400, 1300, 1200]
precipitacion_actual = [1650, 1750, 1550, 1450, 1350, 1550, 1450, 1250, 1150, 1350, 1450, 1250, 1150, 1050]

print("\n COMPARATIVO PRECIPITACIÓN (mm/año):")
print("Departamento          | Promedio | Actual  | Desviación | Estado")
print("-" * 65)

for i in range(len(departamentos)):
    desviacion = precipitacion_actual[i] - precipitacion_promedio[i]
    if desviacion < -300:
        estado = " SEQUÍA SEVERA"
    elif desviacion < -150:
        estado = "  SEQUÍA MODERADA"
    elif desviacion < -50:
        estado = " SEQUÍA LEVE"
    else:
        estado = "✅ NORMAL"

    print(f"{departamentos[i]:<20} | {precipitacion_promedio[i]:>6}   | {precipitacion_actual[i]:>6} | {desviacion:>9}  | {estado}")
```

```python
# Análisis de impacto en cultivos de importancia Economica
print("\n ANÁLISIS DE IMPACTO EN CULTIVOS PRINCIPALES:")
cultivos = {
    "Maíz": {"agua_min": 500, "agua_optima": 800, "tolerancia_sequia": "Media"},
    "Frijol": {"agua_min": 400, "agua_optima": 600, "tolerancia_sequia": "Baja"},
    "Sorgo": {"agua_min": 350, "agua_optima": 550, "tolerancia_sequia": "Alta"},
    "Café": {"agua_min": 1000, "agua_optima": 1600, "tolerancia_sequia": "Baja"}
}

print("\n📋 REQUERIMIENTOS HÍDRICOS POR CULTIVO (mm/temporada):")
print("Cultivo  | Mínimo | Óptimo | Tolerancia Sequía")
print("-" * 50)
for cultivo, datos in cultivos.items():
    print(f"{cultivo:<8} | {datos['agua_min']:>6} | {datos['agua_optima']:>6} | {datos['tolerancia_sequia']:>16}")
```

```python
# Análisis para los 3 departamentos más afectados
print("\n🔍 ANÁLISIS DETALLADO - DEPARTAMENTOS CRÍTICOS:")
departamentos_criticos = ["La Unión", "Morazán", "Cabañas"]

for depto in departamentos_criticos:
    idx = departamentos.index(depto)
    precip = precipitacion_actual[idx]

    print(f"\n📍 {depto.upper()} - Precipitación: {precip} mm/año")
    print("Cultivo     | Riesgo  | Impacto Esperado")
    print("-" * 45)

    for cultivo in cultivos:
        riesgo, impacto = calcular_riesgo(precip, cultivo)
        print(f"{cultivo:<10} | {riesgo:<7} | {impacto}")
```

```python
def calcular_riesgo(precipitacion, cultivo_nombre):
    """
    Calcula el nivel de riesgo y el impacto esperado para un cultivo
    basado en la precipitación y los requerimientos del cultivo.
    """
    if cultivo_nombre not in cultivos:
        return "Desconocido", "N/A"

    datos_cultivo = cultivos[cultivo_nombre]
    agua_min = datos_cultivo["agua_min"]
    tolerancia = datos_cultivo["tolerancia_sequia"]

    if precipitacion < agua_min:
        if tolerancia == "Baja":
            riesgo = "Alto"
            impacto = "Pérdida significativa de cosecha"
        elif tolerancia == "Media":
            riesgo = "Alto"
            impacto = "Pérdida moderada a alta de cosecha"
        else: # Tolerancia Alta
            riesgo = "Medio"
            impacto = "Pérdida leve a moderada de cosecha"
    elif precipitacion < datos_cultivo["agua_optima"] * 0.8:#considera el 80% de riesgo
         if tolerancia == "Baja":
            riesgo = "Medio"
            impacto = "Impacto moderado en rendimiento"
         elif tolerancia == "Media":
            riesgo = "Medio"
            impacto = "Impacto leve a moderado en rendimiento"
         else: # Tolerancia Alta
            riesgo = "Bajo"
            impacto = "Impacto mínimo en rendimiento"
    else:
        riesgo = "Bajo"
        impacto = "Rendimiento esperado normal"

    return riesgo, impacto
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Definir una precipitación de ejemplo
precipitacion_ejemplo = 1100 # mm/año

# Calcular el riesgo para cada cultivo con la precipitación de ejemplo
cultivos_nombres = list(cultivos.keys())
riesgos = [calcular_riesgo(precipitacion_ejemplo, cultivo)[0] for cultivo in cultivos_nombres]

# Mapear niveles de riesgo a un orden para el gráfico (opcional, para ordenar)
orden_riesgo = {"Alto": 3, "Medio": 2, "Bajo": 1, "Desconocido": 0}
riesgos_ordenados = [orden_riesgo[riesgo] for riesgo in riesgos]

# Mapear niveles de riesgo a colores
colores_riesgo = {"Alto": 'darkred', "Medio": 'orange', "Bajo": 'green', "Desconocido": 'gray'}
colores_barras = [colores_riesgo[riesgo] for riesgo in riesgos]

# Crear el gráfico de barras
plt.figure(figsize=(10, 6))
bars = plt.bar(cultivos_nombres, riesgos_ordenados, color=colores_barras)

# Configurar el gráfico
plt.ylabel('Nivel de Riesgo (escala)')
plt.title(f'Nivel de Riesgo por Cultivo con {precipitacion_ejemplo} mm/año', fontweight='bold')
plt.xticks(rotation=0)
plt.yticks([0, 1, 2, 3], ["Desconocido", "Bajo", "Medio", "Alto"])
plt.ylim(0, 3.5) # Ajustar límite y para mejor visualización
plt.grid(axis='y', alpha=0.7)

# Añadir etiquetas de texto con el nivel de riesgo
for bar, riesgo_texto in zip(bars, riesgos):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval, riesgo_texto, va='bottom', ha='center') # va: vertical alignment, ha: horizontal alignment


plt.tight_layout()
plt.show()
```

```python
# Estimacion de Pérdidas Económicas
print("\n ESTIMACIÓN DE PÉRDIDAS ECONÓMICAS:")

hectareas_maiz = [4500, 5200, 3800, 3200, 2800, 3500, 3000, 2500, 2200, 2700, 3300, 2900, 2400, 2000]
rendimiento_normal = 2.72  # toneladas/hectárea
precio_tonelada = 300    # dólares

perdida_total = 0
print("\n Cálculo de pérdidas en cultivo de maíz:")
print("Departamento     | Has Maíz | Pérdida Estimada")
print("-" * 50)

perdidas_totales = {}
perdidas_totales['Maíz'] = 0 # Initialize loss for Maize

for i in range(len(departamentos)):
    if precipitacion_actual[i] < 1400:  # Zonas críticas (assuming < 1400mm/year indicates critical zone for maize based on previous analysis)
        # Assuming a 40% loss in yield for critical zones
        perdida_depto = hectareas_maiz[i] * rendimiento_normal * 0.4 * precio_tonelada
        perdida_total += perdida_depto
        perdidas_totales['Maíz'] += perdida_depto # Accumulate loss for Maize
        print(f"{departamentos[i]:<16} | {hectareas_maiz[i]:>8} | ${perdida_depto:,.0f}")

# Add placeholder losses for other crops for demonstration
# In a real scenario, this would be calculated based on water requirements and actual precipitation
perdidas_totales['Frijol'] = perdida_total * 0.6 # Placeholder
perdidas_totales['Sorgo'] = perdida_total * 0.2 # Placeholder
perdidas_totales['Café'] = perdida_total * 0.8 # Placeholder


print(f"\n PÉRDIDA TOTAL ESTIMADA: ${perdida_total:,.0f}")
```
